```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

# Dataset
data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain',
                'Rain', 'Overcast', 'Sunny', 'Sunny', 'Rain',
                'Sunny', 'Overcast', 'Overcast', 'Rain'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool',
                    'Cool', 'Cool', 'Mild', 'Cool', 'Mild',
                    'Mild', 'Mild', 'Hot', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal',
                 'Normal', 'Normal', 'High', 'Normal', 'Normal',
                 'Normal', 'High', 'Normal', 'High'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak',
             'Strong', 'Strong', 'Weak', 'Weak', 'Weak',
             'Strong', 'Strong', 'Weak', 'Strong'],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes',
             'No', 'Yes', 'No', 'Yes', 'Yes',
             'Yes', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)

# Convert categorical values into numbers
X = pd.get_dummies(df.drop('Play', axis=1))
y = df['Play']

# Build the ID3 decision tree using entropy
model = DecisionTreeClassifier(
    criterion='entropy',
    random_state=0
)
model.fit(X, y)

# Display the decision tree
print("Decision Tree:")
print(export_text(model, feature_names=list(X.columns)))

# Classify a new sample
new_sample = pd.DataFrame([{
    'Outlook': 'Sunny',
    'Temperature': 'Cool',
    'Humidity': 'Normal',
    'Wind': 'Strong'
}])

new_sample = pd.get_dummies(new_sample)
new_sample = new_sample.reindex(columns=X.columns, fill_value=0)

prediction = model.predict(new_sample)

print("\nNew Sample:", new_sample.to_dict(orient='records'))
print("Predicted Class:", prediction[0])
```

Build an Artificial Neural Network by implementing the
Backpropagation algorithm and test the same using appropriate data set

```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Dataset: XOR
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 1, 1, 0])

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=1
)

# Build ANN using backpropagation
model = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='logistic',
    solver='lbfgs',
    max_iter=5000,
    random_state=1
)

# Train the model
model.fit(X_train, y_train)

# Test the model
y_pred = model.predict(X_test)

print("Actual Output:", y_test)
print("Predicted Output:", y_pred)
print("Accuracy:", accuracy_score(y_test, y_pred) * 100, "%")

# Test all XOR inputs
print("\nXOR Predictions:")
for sample in X:
    prediction = model.predict([sample])
    print(sample, "=", prediction[0])
```
