For a given set of training data examples stored in a .CSV file, implement
and demonstrate the Candidate-Elimination algorithm in python to output
a description of the set of all hypotheses consistent with the training
examples# New Section

```python

import pandas as pd

# Step 1: Load the CSV dataset
data = pd.read_csv('/content/findsalgorithm.csv')
data.columns = data.columns.str.strip()

# Step 2: Identify target and features
target = 'EnjoySport'
features = [col for col in data.columns if col != target]

X = data[features].astype(str).values
y = data[target].astype(str).str.strip().str.lower().values

# Step 3: Initialize S and G boundaries
n = len(features)
S = [['Ø'] * n]
G = [['?'] * n]

# Check whether a hypothesis covers an example
def covers(h, x):
    return all(h[j] == '?' or h[j] == x[j]
               for j in range(len(h)))

# Check whether h1 is more general than h2
def more_general(h1, h2):
    return all(a == '?' or a == b
               for a, b in zip(h1, h2))

# Remove duplicate hypotheses
def unique(hypotheses):
    result = []
    for h in hypotheses:
        if h not in result:
            result.append(h)
    return result

# Step 4: Candidate-Elimination algorithm
for x, label in zip(X, y):

    if label == 'yes':
        # Remove G hypotheses that do not cover positive example
        G = [g for g in G if covers(g, x)]

        new_S = []
        for s in S:
            if covers(s, x):
                new_S.append(s)
            else:
                h = s.copy()
                for j in range(n):
                    if h[j] == 'Ø':
                        h[j] = x[j]
                    elif h[j] != x[j]:
                        h[j] = '?'
                new_S.append(h)

        S = unique(new_S)

        # Keep the most specific S hypotheses
        S = [s for s in S if not any(
            s != other and more_general(s, other)
            for other in S
        )]

    else:
        # Remove S hypotheses that cover negative example
        S = [s for s in S if not covers(s, x)]

        new_G = []
        for g in G:
            if not covers(g, x):
                new_G.append(g)
            else:
                # Minimally specialize G using S boundary
                for s in S:
                    for j in range(n):
                        if g[j] == '?' and s[j] not in ['?', 'Ø'] \
                                and s[j] != x[j]:
                            h = g.copy()
                            h[j] = s[j]
                            if any(more_general(h, s2) for s2 in S):
                                new_G.append(h)

        G = unique(new_G)

        # Keep the most general G hypotheses
        G = [g for g in G if not any(
            g != other and more_general(other, g)
            for other in G
        )]

# Step 5: Display results
print("Training Dataset:")
print(data)

print("\nSpecific Boundary (S):")
for h in S:
    print(h)

print("\nGeneral Boundary (G):")
for h in G:
    print(h)

print("\nFeatures:", features)
```
